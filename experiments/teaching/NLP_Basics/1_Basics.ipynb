```python
word = "नमस्ते संसार!"
print(word)
```

```python
print("नमस्ते संसार!")
```

```python
print(word[0])
print(word[1])
print(word[2])
print(word[3])
print(word[4])
print(word[5])
```

```python
# Print the length of the word
print(len(word))

# Count the occurrences of 'स' in the word
print(word.count("स"))

# Find the index of 'स' in the word
print(word.find("स"))

# Find the index of 'संसार' in the word
print(word.index("संसार"))
```

```python
# # Define the string
# word = "Bonjour World!"

# Convert all characters to uppercase
upper_word = word.upper()
print(upper_word)  # Output: 'BONJOUR WORLD!'

# Convert all characters to lowercase
lower_word = word.lower()
print(lower_word)  # Output: 'bonjour world!'

# Capitalize the first letter of every word
title_word = word.title()
print(title_word)  # Output: 'Bonjour World!'

# Capitalize only the first letter of the string
capitalize_word = word.capitalize()
print(capitalize_word)  # Output: 'Bonjour world!'

# Swap the case of all characters
swapcase_word = word.swapcase()
print(swapcase_word)  # Output: 'bONJOUR wORLD!'
```

```python
# Adding "नमस्ते" to the end of the word
print(word + " नमस्ते!")  # Output: 'नमस्ते संसार! नमस्ते!'

# Multiplying the word
print("नमस्ते " * 5)  # Output: 'नमस्ते नमस्ते नमस्ते नमस्ते नमस्ते '

# Reversing the word
print(''.join(reversed(word)))  # Output: '!रास्नस ेतमसन'

# Adding an extra whitespace between every character
print(" ".join(word))  # Output: 'न म स ् त े   स ं स ा र !'
```

```python
# Check if all characters are alphanumeric
print(word.isalnum())  # Output: False

# Check if all characters in the string are alphabetic
print(word.isalpha())  # Output: False

# Check if the string contains digits
print(word.isdigit())  # Output: False

# Check if the string contains title words
print(word.istitle())  # Output: False

# Check if the string is in upper case
print(word.isupper())  # Output: False

# Check if the string is in lower case
print(word.islower())  # Output: False
```

```python
# Check if the string contains only spaces
print(word.isspace())  # Output: False

# Check if the string ends with 'f'
print(word.endswith('f'))  # Output: False

# Check if the string starts with 'H'
print(word.startswith('H'))  # Output: False

# Replace "संसार" with "पिज्जा"
print(word.replace("संसार", "पिज्जा"))  # Output: 'नमस्ते पिज्जा!'

# Slicing the string from 7th to 16th character
print(word[7:16])  # Output: 'संसार!'

# Slicing the string to get the first word
print(word[:7])  # Output: 'नमस्ते'
```
