```python
import fasttext

# 1. Load your model
model = fasttext.load_model("model.bin")

# 2. Access the full vocabulary
words = model.get_words()  # list of strings :contentReference[oaicite:1]{index=1}

# 3. Save them to a text file
with open("vocabulary.txt", "w", encoding="utf-8") as f:
    for w in words:
        f.write(w + "\n")

print(f"✅ Saved {len(words)} tokens to vocabulary.txt")
```
