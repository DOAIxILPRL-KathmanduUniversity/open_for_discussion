```python
# some_file.py
import sys

from Nepali_tokenizer import Tokenizer

# Define the input text
text = ''''त्यतिबेला निर्देशक भाम र नवाजुद्दिनबीच सँगै काम गर्ने कुरा पनि भएको थियो । प्रिमियरमा नवाजुद्दिन र निर्देशक कश्यप विशेष अतिथिका रुपमा सहभागी हुनेछन्,'मिन्सोले भने ।

बर्लिनको प्रमुख प्रतिस्पर्धातर्फ छनौट हुने पहिलो फिल्मको इतिहास रचेको 'साम्बाला' नेपालमा भने भदौ २८ देखि प्रदर्शनमा आउँदैछ ।

यो फिल्मकी प्रमुख अभिनेत्री थिन्ले ल्हामोले हालै लोकार्नो फिल्म फेस्टिभलमा उत्कृष्ट अभिनयका लागि 'बोक्कालिनो डिओर' अवार्ड पाएकी थिइन् ।

'''

# Initialize the tokenizer
tokenizer = Tokenizer()

# Perform various types of tokenization and print the results

# Sentence tokenization
sentences = tokenizer.sentence_tokenize(text)
print("Sentence Tokenization:", sentences)

# Word tokenization
words = tokenizer.word_tokenize(text)
print("Word Tokenization:", words)



# SentencePiece (BPE) tokenization
bpe_tokens = tokenizer.sentencepeice_tokenize(text)
print(len(text.split()))
print("SentencePiece (BPE) Tokenization:", bpe_tokens)
print("SentencePiece (BPE) len:", len(bpe_tokens))

```
