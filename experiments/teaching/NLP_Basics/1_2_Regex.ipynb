```python
# Create the document with email addresses
doc = "थप जानकारीको लागि कृपया हामीलाई इमेल गर्नुहोस्: xyz@abc.com, pqr@mno.com"

# Extract email addresses using re.findall
addresses = re.findall(r'[\w\.-]+@[\w\.-]+', doc)

# Print the extracted email addresses
for address in addresses:
    print(address)
```

```python
new_doc = re.sub(r'([\w\.-]+)@([\w\.-]+)',
r'<email>', doc)
print(new_doc)
```

```python
len(re.findall(r'<email>', new_doc))
```

```python
processed_doc = re.sub(r'\sइमेल\s', " Email ", new_doc)
processed_doc
```

```python
matches = re.findall(r'[\u0900-\u097F]*:', doc)
print(matches)
```

```python
nepali_sentence = 'म यो पुस्तक मन पराउँछु।'


# Find the index of the first instance of 'पुस्तक'
first_index = nepali_sentence.find('पुस्तक')
print(f"First index of 'पुस्तक': {first_index} (-1 if not found)")

# Find the index of the last instance of 'पुस्तक'
last_index = nepali_sentence.rfind('पुस्तक')
print(f"Last index of 'पुस्तक': {last_index} (-1 if not found)")

# Find the index of 'पुस्तक' using index (raises ValueError if not found)
try:
    index_value = nepali_sentence.index('पुस्तक')
    print(f"Index of 'पुस्तक': {index_value} it raises ValueError if not found")
except ValueError:
    print("'पुस्तक' not found")

# Find the last index of 'पुस्तक' using rindex (raises ValueError if not found)
try:
    rindex_value = nepali_sentence.rindex('पुस्तक')
    print(f"Last index of 'पुस्तक': {rindex_value} it raises ValueError if not found")
except ValueError:
    print("'पुस्तक' not found")

# Join the words with a space
joined_sentence = ' '.join(['म', 'यो', 'पुस्तक', 'मन', 'पराउँछु।'])
print(f"Joined sentence: {joined_sentence} ")

# Split the sentence into words
split_words = nepali_sentence.split(' ')
print(f"Split words: {split_words}(whitespace by default)")

# Split the sentence into lines (better demo with multiple lines)
multi_line_sentence = 'म यो पुस्तक मन पराउँछु।\nयो अर्को लाइन हो।'
split_lines_demo = multi_line_sentence.splitlines()
print(f"Split lines demo: {split_lines_demo}")

# Convert to lowercase
lowercase_sentence = nepali_sentence.lower()
print(f"Lowercase sentence: {lowercase_sentence}")

# Convert to uppercase
uppercase_sentence = nepali_sentence.upper()
print(f"Uppercase sentence: {uppercase_sentence}")

# Convert to titlecase
titlecase_sentence = nepali_sentence.title()
print(f"Titlecase sentence: {titlecase_sentence}")

# Strip leading and trailing whitespace
stripped_sentence = nepali_sentence.strip()
print(f"Stripped sentence: {stripped_sentence}")

# Replace 'पुस्तक' with 'किताब'
replaced_sentence = nepali_sentence.replace('पुस्तक', 'किताब')
print(f"Replaced sentence: {replaced_sentence}")
```
