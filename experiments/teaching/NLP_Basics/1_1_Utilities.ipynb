```python
import pandas as pd

text = ['यो NLP को परिचय हो ', 'यो सम्भवतः मानिसहरूका लागि उपयोगी हुनेछ', 'मेसिन लर्निंग नयाँ बिजुली हो', 'AI को वरिपरि कम प्रचार हुनेछ र अगाडि बढी कार्य हुनेछ', 'पाइथन उत्तम उपकरण हो!', 'R राम्रो भाषा हो', 'मलाई यो पुस्तक मन पर्छ', 'मलाई यस जस्तो थप पुस्तकहरू चाहिन्छ']
# convert list to data frame
df = pd.DataFrame({'tweet': text})
print(df)
```

```python
#loweरcasing
df['tweet'] = df['tweet'].apply(lambda x: " ".join(x.lower() for x in x.split()))
df['tweet']
```

```python
df['tweet'] = df['tweet'].str.replace('[^\w\s]', '')
df['tweet']
```

```python
string.punctuation.__add__('।')
s = "यो पुस्तक मलाई मन पर्छ!"
for c in string.punctuation:
    s = s.replace(c, "")
print(s)
```
