```python
from spylls.hunspell import Dictionary
```

```python
import re
import unicodedata

def remove_symbols_and_punctuation(text):
    text = str(text)
    text = unicodedata.normalize('NFKC', text)

    # Remove symbols and punctuation, but retain word characters (letters, numbers, and spaces)
    cleaned_text = re.sub(r'[^\w\s\u0900-\u097F\u0964]', '', str(text))

    return cleaned_text

def dictionary_lookup( word):


    if word:
        lookup = dictionary.lookuper
        return lookup(word)
    else:
        raise ValueError(f"Unsupported")

def check_sentence( sentence, aspercentage = False):
    cleaned_sentence = remove_symbols_and_punctuation(sentence)
    print(cleaned_sentence)
    words = cleaned_sentence.split()
    results = []

    for word in words:
        try:
          if dictionary_lookup( word):
              results.append(1)
          else:
              results.append(0)

        except ValueError as e:
            print(f"Error: {e}")
            results.append(0)

    if aspercentage == True:
      if len(results) == 0:
        return 0
      return sum(results) / len(results) * 100
    else:
      return results

    return results

dictionaries = {
        'nepali': '/content/chipsaldevnagarientry/dictionaries/ne_NP/ne_NP',
        'hindi': '/content/chipsaldevnagarientry/dictionaries/hi_IN/hi_IN',
        'marathi': '/content/chipsaldevnagarientry/dictionaries/mr_IN/mr_IN',
        'bhojpuri': '/content/chipsaldevnagarientry/dictionaries/bho/bho',
        'sanskrit': '/content/chipsaldevnagarientry/dictionaries/sa_Deva/sa-Deva'

    }


```

```python
dictionary = Dictionary.from_files(dictionaries["nepali"])


sentence = "यो एउटा परीक्षण वाक्य हो।"
result = check_sentence(sentence,aspercentage = True)
print(result)

lookup = dictionary.lookuper

for form in lookup.good_forms('बुधबारको'):
    print(form)
```
