```python
import csv
import wn

# Download the English WordNet if not already available
if not wn.download('ewn:2020'):
    print("Failed to download English WordNet. Please check your internet connection.")
    exit(1)

# Load the English WordNet
ewn = wn.Wordnet('ewn:2020')

# Prepare CSV data
csv_data = [['ILI', 'Lemmas', 'Part of Speech', 'Definition', 'Examples']]

# Iterate through all synsets in the English WordNet
for synset in ewn.synsets():
    ili = synset.ili
    lemmas = ', '.join([sense.word().lemma() for sense in synset.senses()])
    pos = synset.pos
    definition = synset.definition()
    examples = '; '.join(synset.examples()) if synset.examples() else "No examples"

    csv_data.append([ili, lemmas, pos, definition, examples])

# Write CSV file
output_file = 'wordnet_ili_data.csv'
with open(output_file, 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(csv_data)

print(f"CSV file '{output_file}' created successfully.")
print(f"Total synsets processed: {len(csv_data) - 1}")
```
