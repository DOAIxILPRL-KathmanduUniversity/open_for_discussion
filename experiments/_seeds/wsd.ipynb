```python
import pyiwn
import re
from collections import Counter
```

```python
# नेपाली भाषा इनडोवर्डनेट इन्टरफेस आरम्भ गर्नुहोस्
iwn = pyiwn.IndoWordNet(lang=pyiwn.Language.NEPALI)
```

```python
# उदाहरण वाक्यहरू
sentences = [
    "उसले आफ्नो अंशको हिस्सा पाएको छ।",
    "तपाईंको अंशको मूल्य के हो?",
    "उसको अंशको मूल्यांकन राम्रो छ।",
    "उसको अंशको मूल्यांकन राम्रो छ।"
]
```

```python
def get_glosses(word):
    """
    दिइएको शब्दका सबै साइनसेटहरूको अर्थ (gloss) प्राप्त गर्नुहोस्।
    """
    try:
        synsets = iwn.synsets(word)
        glosses = [synset.gloss() for synset in synsets]
        return glosses
    except:
        return ['']
```

```python
def lesk_algorithm(sentence, word):
    """
    लेक्स एल्गोरिदम लागू गरेर शब्दको अर्थ पहिचान गर्नुहोस्।
    """
    context = re.findall(r'\w+', sentence)
    context = [w for w in context if w != word]
    glosses = get_glosses(word)
    max_overlap = 0
    best_sense = None
    for gloss in glosses:
        gloss_tokens = set(re.findall(r'\w+', gloss))
        overlap = len(set(context) & gloss_tokens)
        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = gloss
    return best_sense
```

```python
for sentence in sentences:
    words = sentence.split(' ')
    for word in words:
        sense = lesk_algorithm(sentence, word)
        print(f"शब्द: {word}")
        print(f"अर्थ: {sense}")
        print("-" * 50)
```
